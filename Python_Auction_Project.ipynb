```python
def place_bid():
    user_bid = float(input("Enter your bid: "))
    return user_bid


def tell_user_to_pay(item_name):
    print("You won the " + item_name + "!")
    print("You need to pay within 24 hours.")


item_name = input("Enter the name of the item: ")
winning_bid = float(input("Enter the starting bid: "))
print(item_name + " has been listed.")

user_bid = place_bid()

while user_bid <= winning_bid:
    print("Your bid is too low. Please try again.")
    user_bid = place_bid()

tell_user_to_pay(item_name)
```
